```python
def local_example():
    local_var = "이것은 지역 변수입니다"
    print(local_var)  # 함수 내에서 접근 가능

# 전역 변수 예시
global_var = "이것은 전역 변수입니다"
def global_example():
    print(global_var)  # 전역 변수는 함수 내에서도 접근 가능
# 정적 변수 예시
class Example:
    static_var = "이것은 정적 변수입니다"  # 클래스 변수(정적변수) 
    def print_static():
        print(Example.static_var)  # 클래스를 통해 접근 가능

```

```python
counter = 0  # 전역 변수
def increment_counter():
    global counter  # global 키워드로 전역 변수 사용 선언
    counter += 1
    print(f"카운터: {counter}")
increment_counter()  # 출력: 카운터: 1
increment_counter()  # 출력: 카운터: 2
increment_counter()  # 출력: 카운터: 3
```

```python
orders = [
{'order_id': 'ORD001', 'product': '노트북', 'price': 
1500000, 'qty': 2},{'order_id': 'ORD002', 'product': '마우스', 'price': 
30000, 'qty': 5},
{'order_id': 'ORD003', 'product': '키보드', 'price': 
120000, 'qty': 3},
{'order_id': 'ORD004', 'product': '모니터', 'price': 
400000, 'qty': 1}
]
print("=== 주문 내역 ===")
total_amount = 0
for order in orders:
    subtotal = order['price'] * order['qty']
    total_amount += subtotal
    print(f"{order['order_id']}: {order['product']} x {order['qty']}개 = {subtotal:,}원")
print(f"\n총 주문 금액: {total_amount:,}원")
```
