```python
sentence = input("문장을 입력하세요: ")
words = ["파이썬", "코딩", "데이터", "AI"]
found_word = None

for word in words:
    if word in sentence:
        found_word = word
        break

if found_word:
    print(f"문장에서 '{found_word}'를 찾았습니다.")

else:
    print("입력한 문장에 해당 단어가 없습니다.")
```
