```python
class Calculator:

  def __init__(self, number_list):
    self.number_list = number_list

  def sum(self):
    return sum(self.number_list)

  def avg(self):
    return sum(self.number_list) / len(self.number_list)


# 실행 예시
cal1 = Calculator([1, 2, 3, 4, 5])
print(cal1.sum())  # 15
print(cal1.avg())  # 3.0

cal2 = Calculator([6, 7, 8, 9, 10])
print(cal2.sum())  # 40
print(cal2.avg())  # 8.0
```

```python
# Calculator 클래스 정의 및 호출
class Calculator:
    def __init__(self):
        self.value = 0
        
    def add(self, val):
        self.value += val
        return self.value

# 객체 생성
calc = Calculator()
print(calc.add(10)) # 호출

# Account 클래스 정의 및 호출
class Account:
    def __init__(self, balance):
        self.balance = balance
        
    def deposit(self, amount):
        self.balance += amount
        return self.balance

# 객체 생성
my_account = Account(1000)
print(my_account.deposit(500))
```

### 객체의 연산자
## 객체지향

```python
# 객체지향의 특징(예: 사칙연산 클래스 연산자 활용 등) 중 간단한 예시
class Number:
    def __init__(self, val):
        self.val = val
        
    # 특수 메서드(연산자 오버로딩) 예시
    def __add__(self, other):
        return self.val + other.val

n1 = Number(10)
n2 = Number(20)
print(n1 + n2) # 객체끼리의 연산 (+)
```

# 예외클래스 정의


```python
class AgeException(Exception):
    def __init__(self, msg):
        self.message = msg
```

```python
def input_age():
    try:
        age = int(input("나이를 입력하세요: "))
   
        if age <= 0:
            raise AgeException('나이는 0보다 커야합니다')
        elif age > 100:
            raise AgeException('진짜?')
        else:
            return age
        
    except AgeException as ae:
        print(f"에러 발생: {ae}")
input_age()
```
