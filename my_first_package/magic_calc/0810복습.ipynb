```python
fruits = ["사과", "바나나", "포도"]
for fruit in fruits:
    print(fruit)

for i in range(3): #range(a,b,step) 이때 b는 포함하지 않는다.
    print(i)

orders = [
{'order_id': 'ORD001', 'product': '노트북', 'price': 
1500000, 'qty': 2},
{'order_id': 'ORD002', 'product': '마우스', 'price': 
30000, 'qty': 5},
{'order_id': 'ORD003', 'product': '키보드', 'price': 
120000, 'qty': 3},
{'order_id': 'ORD004', 'product': '모니터', 'price': 
400000, 'qty': 1}]

print("=== 주문 내역 ===")

total_amount = 0
for order in orders:
    subtotal = order['price'] * order['qty']
    total_amount += subtotal
print(f"{order['order_id']}: {order['product']} x {order['qty']}개 = {subtotal:,}원")
print(f"\n총 주문 금액: {total_amount:,}원")
```

```python
password = "98202400"
has_upper = False
has_lower = False
has_digit = False
has_special = False

for char in password:
    if char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
    elif char.isdigit():
        has_digit = True
    elif char in "!@#$%^&*":
        has_special = True
print("=== 비밀번호 강도 분석 ===")
print(f"대문자 포함: {'✅' if has_upper else '❌'}")
print(f"소문자 포함: {'✅' if has_lower else '❌'}")
print(f"숫자 포함: {'✅' if has_digit else '❌'}")
print(f"특수문자 포함: {'✅' if has_special else '❌'}")

if has_upper and has_lower and has_digit and has_special:
    print("\n비밀번호 강도: 강함")
else:
    print("\n비밀번호 강도: 약함")
```
