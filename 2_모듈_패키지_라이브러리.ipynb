```python
import pandas as pd #라이브러리 호출

#표로 보기위한 원시데이터(딕셔너리)
data = {
    '이름': ['김철수', '이영희', '박민수', '정수진', '최영호', '한미래', '윤지원'],
    '나이': [25, 28, 22, 31, 27, 24, 29],
    '점수': [85, 92, 78, 95, 88, 73, 91]}
```

```python
df = pd.DataFrame(data) #판다스 데이터
df.head()
```

```python
import math

average_age = math.trunc(df['나이'].mean())
# 또는 소수점 이하 내림을 원할 경우: df['나이'].mean() // 1
print(average_age)
```
